# 4.1. To perform Term-Frequency Analysis and Named Entity Recognition on a given text using any toolkit.
1. Calculate the frequency of each word and save the output in CSV format (Term, Frequency)
2. Identify the top 10 most frequent terms and display their corresponding frequencies.



```python
import nltk
import re
import csv
from collections import Counter
from nltk.tokenize import word_tokenize
from nltk import ne_chunk, pos_tag
from nltk.tree import Tree
```

```python
# Download required resources
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('maxent_ne_chunker_tab')
nltk.download('words')
```

```python
# Read file
with open("/content/drive/MyDrive/CLNLP-Lab/Experiment-4/4.1_4.2_input.txt", "r", encoding="utf-8") as file:
    text = file.read()
```

```python
# Convert to lowercase
text_lower = text.lower()
```

```python
# Tokenize
words = word_tokenize(text_lower)
```

```python
# Keep only words
words = [word for word in words if word.isalpha()]
```

```python
# Calculate frequency
frequency = Counter(words)
```

```python
# Save frequency to CSV
with open("term_frequency_4_1.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["Term", "Frequency"])

    for term, freq in frequency.items():
        writer.writerow([term, freq])
```

```python
# Display top 10 terms
print("Top 10 Most Frequent Terms:")
for term, freq in frequency.most_common(10):
    print(term, ":", freq)
```

```python
# Named Entity Recognition
original_words = word_tokenize(text)
pos_tags = pos_tag(original_words)

named_entities = ne_chunk(pos_tags)

print("\nNamed Entities:")

for chunk in named_entities:
    if isinstance(chunk, Tree):
        entity = " ".join(word for word, tag in chunk.leaves())
        entity_type = chunk.label()
        print(entity, "->", entity_type)
```

# 4.2. To perform Term-Frequency Analysis on a given text without using any toolkit.

1. Calculate the frequency of each word and save the output in CSV format (Term, Frequency)
2. Identify the top 10 most frequent terms and display their corresponding frequencies.


```python
import re
import csv
from collections import Counter
```

```python
with open("/content/drive/MyDrive/CLNLP-Lab/Experiment-4/4.1_4.2_input.txt", "r", encoding="utf-8") as file:
    text = file.read()
```

```python
text = text.lower()
```

```python
words = re.findall(r'\b[a-z]+\b', text)
```

```python
frequency = Counter(words)
```

```python
with open("term_frequency_without_toolkit.csv",
          "w",
          newline="",
          encoding="utf-8") as file:

    writer = csv.writer(file)

    writer.writerow(["Term", "Frequency"])

    for term, count in frequency.most_common():
        writer.writerow([term, count])

print("Term frequency saved to "
      "term_frequency_without_toolkit.csv")
```

```python
print("\nTop 10 Most Frequent Terms:")

for term, count in frequency.most_common(10):
    print(term, ":", count)
```

# 4.3. Write a Python program without using NLTK, spaCy, or any other NLP toolkit to:
1. Read multiple text documents.
2. Convert text to lowercase.
3. Remove punctuation.
4. Tokenize the documents.
5. Calculate TF for every term.
6. Calculate DF for every term.
7. Calculate IDF using the logarithmic formula.
8. Calculate TF-IDF for every term in every document.
9. Display Term, TF, DF, IDF, and TF-IDF at each step.
10. Identify the top 10 terms with the highest TF-IDF score for each document.


```python
import re
import math
```

```python
documents = {
    "Document 1": "Natural language processing is a field of artificial intelligence.",
    "Document 2": "Natural language processing helps computers understand human language.",
    "Document 3": "Machine learning is an important part of artificial intelligence."
}
```

```python
processed_documents = {}
for doc_name, text in documents.items():

    # Convert to lowercase
    text = text.lower()

    # Remove punctuation
    text = re.sub(r'[^\w\s]', '', text)

    # Tokenize
    words = text.split()

    processed_documents[doc_name] = words

print("Processed Documents:")
for doc, words in processed_documents.items():
    print(doc, ":", words)
```

```python
tf = {}

for doc, words in processed_documents.items():

    tf[doc] = {}

    total_words = len(words)

    for word in words:

        if word not in tf[doc]:
            tf[doc][word] = 0

        tf[doc][word] += 1

    # Convert count to TF
    for word in tf[doc]:
        tf[doc][word] = tf[doc][word] / total_words


print("\nTF:")
for doc in tf:
    print("\n", doc)

    for term, value in tf[doc].items():
        print(term, ":", round(value, 4))
```

```python
df = {}

for doc, words in processed_documents.items():

    unique_words = set(words)

    for word in unique_words:

        if word not in df:
            df[word] = 0

        df[word] += 1


print("\nDF:")
for term, value in df.items():
    print(term, ":", value)
```

```python
N = len(documents)

idf = {}

for term, df_value in df.items():

    idf[term] = math.log(N / df_value)


print("\nIDF:")

for term, value in idf.items():
    print(term, ":", round(value, 4))
```

```python
tfidf = {}

for doc in tf:

    tfidf[doc] = {}

    for term, tf_value in tf[doc].items():

        tfidf[doc][term] = tf_value * idf[term]


print("\nTF-IDF:")

for doc in tfidf:

    print("\n", doc)

    for term, value in tfidf[doc].items():

        print(
            term,
            "TF =", round(tf[doc][term], 4),
            "DF =", df[term],
            "IDF =", round(idf[term], 4),
            "TF-IDF =", round(value, 4)
        )
```

```python
print("\nTop 10 TF-IDF Terms for Each Document:")

for doc in tfidf:

    sorted_terms = sorted(
        tfidf[doc].items(),
        key=lambda x: x[1],
        reverse=True
    )

    print("\n", doc)

    for term, score in sorted_terms[:10]:

        print(
            term,
            "->",
            round(score, 4)
        )
```
